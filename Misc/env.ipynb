```python
from dotenv import dotenv_values
from pathlib import Path
import json


# Load
cfg = dotenv_values('.env.prac')    # {'S_STR': 'smollm2:135m'}

# Print
pprint(cfg)


# Read - Scalars
s_str       = cfg['S_STR']
s_int       = int(cfg['S_INT'])
s_float     = float(cfg['S_FLOAT'])
s_bool      = json.loads(cfg['S_BOOL'])
s_null      = json.loads(cfg['S_NULL'])

# Read - Collections
col_list    = json.loads(cfg['COL_LIST'])
col_dict    = json.loads(cfg['COL_DICT'])
col_tuple   = tuple(json.loads(cfg['COL_TUPLE']))
col_set     = set(json.loads(cfg['COL_SET']))

# Read - Path
data_dir    = Path(cfg['DATA_DIR']).resolve()
```
