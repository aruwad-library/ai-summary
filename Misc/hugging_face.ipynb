# Memory Usage

```python
from transformers import AutoModel

model_name = 'distilbert-base-uncased'
model = AutoModel.from_pretrained(model_name)

memory_usage = model.get_memory_footprint() / 1024**3   # MB
print(f"Memory usage: {memory_usage:.4f} MB")
```
