# Data Types

## dict

```python
d = {
    'a': 1,
    'b': 2,
}

dd = dict(      # more parameter style.
    a=1,
    b=2,
)

dd.update({
    'a': 100,   # overwrites for the same key.
})
```

# Code Execution

```python
code = """
def say_name(name: str) -> None:
    print(f'My name is {name}.')
"""

namespace = {}

exec(code, namespace)

ftn = namespace['say_name']     # ftn = def say_name(name: str)


try:
    ftn('yana')

except:
    raise(ValueError)
```

# math

```python
from math import isclose

a = 1_000_000
b = 1_000_000.0001

print(a == b)
print(isclose(a, b))
```

# json

```python
import json

d = {
    'name': 'yana',
    'job': 'mle',
}

s = json.dumps(d)

print(type(s))
print(s)

j = json.loads(s)

print(j)
```
