# Wikipedia

```python
import wikipediaapi

wiki = wikipediaapi.Wikipedia(
    user_agent='LangChainRAG/1.0',
    language='en',
)
results = wiki.search('Retrieval-augmented generations', limit=3)

pages = results.pages

for keyword, value in pages.items():
    print(f"[{keyword}]\n{value.text[:100]} ...\n")
```
