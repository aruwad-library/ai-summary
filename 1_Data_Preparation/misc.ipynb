# Random Null

```python
# Sample dataset.
from sklearn.datasets import fetch_openml
df = fetch_openml('titanic', version=1).data
display(df.info())

# Random null, 10%.
rng = np.random.default_rng(42)
ratio = 0.1
df = df.mask(rng.random(df.shape) < ratio)
display(df.info())
```
