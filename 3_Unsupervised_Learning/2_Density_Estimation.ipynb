# 1. Kernel Density Estimation

## 1) Summary

- Predict a probability density based on the assumption of the density distribution of other samples.
- $\hat{p}(x) = \frac{1}{m} \displaystyle\sum_{i=1}^m K_h(x - x_i)$
  - $\hat{p}(x)$ = a probability density of $x$
  - $m$ = a number of samples
  - $K_h$ = a kernel function within $h = \mathrm{bandwidth}$

## 2) API

````python
from sklearn.neighbors import KernelDensity

# Data
X_train = np.array([1.0, 2.0, 4.0]).reshape(-1, 1)

# Model
kde = KernelDensity(
    kernel="gaussian",
    bandwidth=1.0,
)

# Fit
kde.fit(X_train)

# Predict
log_density = kde.score_samples([[3.0]])
density     = np.exp(log_density)
````

# 2. Gaussian Mixture Model

## 1) Summary

- Supposes there are $K$ independent distribution where the samples are from.

1. Initialize $K, \mu_k, \sigma_k$ for $N(\mu_k, \sigma_k)$.
2. Expectation: allocate all $x_i$ according to the maxinum probability.
3. Maximization: adjust $\mu_k, \sigma_k$ according to the samples belonging to $k$.
4. Repeat 2-3 until it converges.

- Prediction
  - $\hat{p}(x) = \displaystyle \sum_{k=1}^K \pi_k \times N(x | \mu_k, \Sigma_k)$, where $\pi_k = \frac1m \times \displaystyle \sum_i \gamma_{ik}$
  - $\hat{p}(z = k | x) = \frac{\pi_k \times N(x | \mu_k, \Sigma_k)}{\displaystyle \sum_{j=1}^K \pi_j \times N(x | \mu_j, \Sigma_j)}$

- Evaluation
  - AIC, Akaike Information Criterion
    - $AIC = -2 \log L + 2p$, where $L$ = likelihood, $p$ = a number of parameters
  - BIC, Bayesian Information Criterion
    - $BIC = -2 \log L + p \log m$, $m$ = a number of samples
